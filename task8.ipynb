```python
for col in tips.select_dtypes(include='number').columns:
    tips[col] = tips[col].fillna(tips[col].mean())

for col in tips.select_dtypes(include='object').columns:
    tips[col] = tips[col].fillna(tips[col].mode()[0])

print(tips.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

tips['sex'] = le.fit_transform(tips['sex'])
tips['smoker'] = le.fit_transform(tips['smoker'])
tips['day'] = le.fit_transform(tips['day'])
tips['time'] = le.fit_transform(tips['time'])

print(tips.head())
```

```python
sns.countplot(x='sex', data=sns.load_dataset('tips'))
plt.title('Count Plot - Sex')
plt.show()
```

```python
sns.countplot(x='smoker', data=sns.load_dataset('tips'))
plt.title('Count Plot - Smoker')
plt.show()
```

```python
sns.countplot(x='day', data=sns.load_dataset('tips'))
plt.title('Count Plot - Day')
plt.show()
```

```python
tips = sns.load_dataset('tips')

print(tips['sex'].value_counts())
print(tips['smoker'].value_counts())
print(tips['day'].value_counts())
print(tips['time'].value_counts())
```

```python
print(tips['sex'].value_counts(normalize=True) * 100)

print(tips['smoker'].value_counts(normalize=True) * 100)

print(tips['day'].value_counts(normalize=True) * 100)

print(tips['time'].value_counts(normalize=True) * 100)
```

```python
print("Most common sex:", tips['sex'].mode()[0])

print("Most common smoker:", tips['smoker'].mode()[0])

print("Most common day:", tips['day'].mode()[0])

print("Most common time:", tips['time'].mode()[0])
```

```python
tips['day'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Day Distribution')
plt.ylabel('')
plt.show()
```

```python
tips['sex'].value_counts().plot(kind='bar')

plt.title('Sex Distribution')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```
